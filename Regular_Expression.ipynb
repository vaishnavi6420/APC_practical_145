```python
# 1. Extract all digits from a string
import re
text = "My roll number is 123 and my marks are 456."
digits = re.findall(r"\d", text)
print("Digits:", digits)
```

```python
# 2. Display words beginning with A or a
import re
sentence = "Alice and an apple are amazing examples."
words = re.findall(r"\b[Aa]\w*", sentence)
print("Words beginning with A/a:", words)
```

```python
# 3. Count occurrences of vowels
import re
text = "Regular expressions are useful."
vowels = re.findall(r"[aeiou]", text, re.IGNORECASE)
print("Vowel occurrences:", len(vowels))
print("Vowels:", vowels)
```

```python
# 4. Extract all uppercase letters
import re
text = "Python REGEX Is Powerful."
uppercase = re.findall(r"[A-Z]", text)
print("Uppercase letters:", uppercase)
```

```python
# 5. Identify spaces, tabs, and newline characters
import re
text = "Hello world.\nPython\tRegex  is useful."
spaces = re.findall(r" ", text)
tabs = re.findall(r"\t", text)
newlines = re.findall(r"\n", text)
print("Spaces:", len(spaces))
print("Tabs:", len(tabs))
print("Newlines:", len(newlines))
```

```python
# 6. Replace multiple consecutive spaces with one space
import re
text = "Python    regular     expressions   are useful."
result = re.sub(r" {2,}", " ", text)
print(result)
```

```python
# 7. Display words containing exactly five characters
import re
sentence = "Python hello world apple chair table."
words = re.findall(r"\b[A-Za-z]{5}\b", sentence)
print("Five-character words:", words)
```

```python
# 8. Identify words ending with the letter s
import re
sentence = "This program uses regular expressions and strings."
words = re.findall(r"\b\w+s\b", sentence, re.IGNORECASE)
print("Words ending with s:", words)
```

```python
# 9. Remove special characters while retaining letters, digits, and spaces
import re
text = "Hello@ Python! 123 #Regex$"
result = re.sub(r"[^A-Za-z0-9 ]", "", text)
print(result)
```

```python
# 10. Count the number of words in a sentence
import re
sentence = "Python regular expressions are very useful."
words = re.findall(r"\b\w+\b", sentence)
print("Number of words:", len(words))
```

```python
# 11. Validate a 10-digit Indian mobile number
import re
mobile = input("Enter Indian mobile number: ")
pattern = r"^[6-9]\d{9}$"
if re.fullmatch(pattern, mobile):
    print("Valid Indian mobile number")
else:
    print("Invalid Indian mobile number")
```

```python
# 12. Validate an email address
import re
email = input("Enter email address: ")
pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
if re.fullmatch(pattern, email):
    print("Valid email address")
else:
    print("Invalid email address")
```

```python
# 13. Extract dates in DD-MM-YYYY format
import re
paragraph = "The exams are on 15-10-2026 and 25-12-2026."
dates = re.findall(r"\b\d{2}-\d{2}-\d{4}\b", paragraph)
print("Dates:", dates)
```

```python
# 14. Identify web addresses beginning with http:// or https://
import re
text = "Visit http://example.com and https://www.google.com for details."
urls = re.findall(r"https?://[^\s]+", text)
print("Web addresses:", urls)
```

```python
# 15. Extract all hashtags from a social media post
import re
post = "Learning #Python and #Regex today! #Programming is fun."
hashtags = re.findall(r"#\w+", post)
print("Hashtags:", hashtags)
```

```python
# 16. Validate a strong password
import re
password = input("Enter password: ")
pattern = r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[^A-Za-z0-9]).{8,}$"
if re.fullmatch(pattern, password):
    print("Strong password")
else:
    print("Password does not meet the requirements")
```

```python
# 17. Identify consecutive repeated words ignoring case
import re
sentence = "This is is a test test of the Regex regex program."
pattern = r"\b(\w+)(?:\s+\1\b)"
matches = re.findall(pattern, sentence, re.IGNORECASE)
print("Consecutive repeated words:", matches)
```

```python
# 19. Validate an Indian vehicle registration number
import re
registration = input("Enter vehicle registration number: ").upper()
pattern = r"^[A-Z]{2}\d{2}[A-Z]{2}\d{4}$"
if re.fullmatch(pattern, registration):
    print("Valid vehicle registration format")
else:
    print("Invalid vehicle registration format")
```

```python
# 21. Remove HTML tags, extra whitespace, and unwanted punctuation
import re
text = "<p>Hello,   Python!</p>\n<div>Regular expressions are useful.</div>"
text = re.sub(r"<[^>]+>", " ", text)
text = re.sub(r"[^A-Za-z0-9\s]", "", text)
text = re.sub(r"\s+", " ", text).strip()
print(text)
```

```python
# 22. Extract invoice numbers, dates, and monetary amounts
import re
invoice_text = """
Invoice No: INV-2026-001
Date: 16-10-2026
Total Amount: Rs. 12,500.00
"""
invoice_numbers = re.findall(r"\bINV-\d{4}-\d+\b", invoice_text)
dates = re.findall(r"\b\d{2}-\d{2}-\d{4}\b", invoice_text)
amounts = re.findall(r"(?:Rs\.?|₹)\s?[\d,]+(?:\.\d{2})?", invoice_text)
print("Invoice numbers:", invoice_numbers)
print("Dates:", dates)
print("Monetary amounts:", amounts)
```

```python
# 23. Identify single-line comments in Python source code
import re
code = """# This is a comment
x = 10  # Inline comment
print("This is # inside a string")
"""
# Simple regex approach for single-line comments.
# Regex alone cannot correctly parse every Python string literal
# or all multiline cases.
comments = re.findall(r"(?m)#.*$", code)
print("Possible single-line comments:")
for comment in comments:
    print(comment)
```

```python
# 24. Validate product codes in ABC-1234-XY format
import re
code = input("Enter product code: ").upper()
pattern = r"^[A-Z]{3}-\d{4}-[A-Z]{2}$"
if re.fullmatch(pattern, code):
    print("Valid product code")
else:
    print("Invalid product code")
```

```python
# 25. Extract student names, roll numbers, and marks
import re
text = """
Name: Rahul Patil, Roll No: 101, Marks: 85
Name: Sneha More, Roll No: 102, Marks: 91
"""
pattern = r"Name:\s*([A-Za-z ]+),\s*Roll No:\s*(\d+),\s*Marks:\s*(\d+)"
students = re.findall(pattern, text)
for name, roll, marks in students:
    print("Name:", name.strip())
    print("Roll No:", roll)
    print("Marks:", marks)
    print()
```

```python
# 26. Extract DOI identifiers, publication years, email addresses, and URLs
import re
references = """
Smith J. (2024). A study of Python. doi:10.1234/example.2024
Contact: author@example.com
URL: https://example.com/paper
Jones A. (2023). Data Analysis. doi:10.5678/abc.2023
"""
dois = re.findall(r"\b10\.\d{4,9}/[-._;()/:A-Za-z0-9]+", references)
years = re.findall(r"\b(?:19|20)\d{2}\b", references)
emails = re.findall(r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b", references)
urls = re.findall(r"https?://[^\s]+", references)
print("DOIs:", dois)
print("Publication years:", years)
print("Email addresses:", emails)
print("URLs:", urls)
```
